## Q.6 Create a sample dataset (employees.csv) containing information about employees in a company.

```python
data = {
    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Edward"],
    "Department": ["HR", "IT", "IT", "Marketing", "Sales"],
    "Age": [29, 34, 41, 28, 38],
    "Salary": [50000, 70000, 65000, 55000, 60000],
    "Years_of_Experience": [4, 8, 10, 3, 12],
    "Joining_Date": ["2020-03-15", "2017-07-19", "2013-06-01", "2021-02-10", "2010-11-25"],
    "Gender": ["Female", "Male", "Male", "Female", "Male"],
    "Bonus": [5000, 7000, 6000, 4500, 5000],
    "Rating": [4.5, 4.0, 3.8, 4.7, 3.5]
}
df = pd.DataFrame(data)
display(df)
```

### Add a new column that categorizes employees based on their performance rating:
Excellent for ratings >= 4.5

Good for ratings >= 4.0 but < 4.5

Average for ratings < 4.0

```python
def categorize_rating(rating):
    if rating >= 4.5:
        return 'Excellent'
    elif rating >= 4.0:
        return 'Good'
    else:
        return 'Average'

df['Performance_Category'] = df['Rating'].apply(categorize_rating)
display(df[['Name', 'Rating', 'Performance_Category']])
```

### Identify missing values in the DataFrame.

```python
missing_values = df.isnull().sum()
display(missing_values)
```

### Rename the 'Employee_ID' column to 'ID'.

```python
df = df.rename(columns={'Employee_ID': 'ID'})
display(df.head())
```

### Find all employees who have more than 5 years of experience.

```python
experienced_employees = df[df['Years_of_Experience'] > 5]
display(experienced_employees[['Name', 'Years_of_Experience']])
```

### Find all employees who belong to the IT department.

```python
it_employees = df[df['Department'] == 'IT']
display(it_employees[['Name', 'Department']])
```

### Modify the dataset by adding a new column, Tax, which deducts 10% of the Salary.

```python
df['Tax'] = df['Salary'] * 0.10
display(df[['Name', 'Salary', 'Tax']].head())
```

### Save the modified DataFrame (with added columns) to a new CSV file.

```python
df.to_csv('employees_modified.csv', index=False)
```
