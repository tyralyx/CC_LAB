Set the random seed to your own roll number using random.seed(your_roll_number) before generating your numbers — this keeps your list random-looking but reproducible and unique to you (a different seed produces a different list, so this cannot be copied from a classmate).

Generate a list of 100 random numbers between 100 and 900 (inclusive).
Count and print all odd numbers in the list.

Count and print all even numbers in the list.

Count and print all prime numbers in the list, and additionally build the actual list of prime numbers found using list comprehension.

Print the number that occurs most frequently in your list, and how many times it occurs.

```python
import random
from collections import Counter

random.seed(roll_no)
print(f"Random seed set to: {roll_no}")

random_numbers = [random.randint(100, 900) for _ in range(100)]
print("Generated list of 100 random numbers (first 10 shown):", random_numbers[:10])
```

```python
odd_numbers_count = sum(1 for num in random_numbers if num % 2 != 0)
print(f"Count of odd numbers: {odd_numbers_count}")

even_numbers_count = sum(1 for num in random_numbers if num % 2 == 0)
print(f"Count of even numbers: {even_numbers_count}")
```

```python
def is_prime(num):
    if num <= 1:
        return False
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            return False
    return True

prime_numbers_list = [num for num in random_numbers if is_prime(num)]
prime_numbers_count = len(prime_numbers_list)

print(f"Count of prime numbers: {prime_numbers_count}")
print("List of prime numbers found:", prime_numbers_list)
```

```python
if random_numbers:
    number_counts = Counter(random_numbers)
    most_common_number, frequency = number_counts.most_common(1)[0]
    print(f"The most frequent number is: {most_common_number}, which occurs {frequency} times.")
else:
    print("The list of random numbers is empty.")
```

Using the same 8 digits of your roll number from Q1 (before multiplying by 10), create set A = {digit × 7 for
each digit} and set B = {digit × 9 for each digit}. Print both sets.

Find the union of A and B — all unique values across both sets.

Find the intersection of A and B — values common to both.

Find A − B and B − A separately (not combined) using difference(), and explain in one line how this
differs from symmetric_difference().

Find the symmetric difference of A and B.

Check whether A is a subset of B, and whether B is a superset of A, using issubset() and issuperset().

Ask the user to input a value X. Remove X from set A if present using discard() (not remove()) — explain
in one line why discard() is safer than remove() when you're not sure the value exists.

```python
roll_no_digits = [int(digit) for digit in str(roll_no)[:8]]

A = {digit * 7 for digit in roll_no_digits}
B = {digit * 9 for digit in roll_no_digits}

print("Set A:", A)
print("Set B:", B)
```

```python
union_AB = A.union(B)
print("Union of A and B (A | B):", union_AB)
```

```python
intersection_AB = A.intersection(B)
print("Intersection of A and B (A & B):", intersection_AB)
```

```python
difference_A_B = A.difference(B)
difference_B_A = B.difference(A)

print("A - B (elements in A but not in B):", difference_A_B)
print("B - A (elements in B but not in A):", difference_B_A)
# The difference() method finds elements unique to one set when compared to another, whereas symmetric_difference() finds elements unique to either set (those not in their intersection).
```

```python
symmetric_difference_AB = A.symmetric_difference(B)
print("Symmetric Difference of A and B (A ^ B):", symmetric_difference_AB)
```

```python
is_A_subset_B = A.issubset(B)
is_B_superset_A = B.issuperset(A)

print(f"Is A a subset of B?: {is_A_subset_B}")
print(f"Is B a superset of A?: {is_B_superset_A}")
```

```python
x_value = int(input("Please enter a value X to attempt to remove from set A: "))

print(f"Set A before discard({x_value}):", A)
A.discard(x_value)
print(f"Set A after discard({x_value}):", A)
```
